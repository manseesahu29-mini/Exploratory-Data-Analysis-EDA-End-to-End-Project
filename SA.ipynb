```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
df =pd.read_csv('cx.csv')
df.head()
```

#replacing the blank with 0 as tenure is 0 and no total charges are recorded . We also chnage the data type from object to FLOAT

```python
df["TotalCharges"]=df["TotalCharges"].replace(" ","0")
df["TotalCharges"]=df["TotalCharges"].astype("float")
```

```python
def conv(value):
    if value==1:
        return "yes"
    else:
        return"no"
df['SeniorCitizen']=df['SeniorCitizen'].apply(conv)
```

#converted 0 and 1 value of senior citizen to yes/no to make it easier to understand

```python
ax= sns.countplot(x='Churn',data=df)
ax.bar_label(ax.containers[0])
plt.title("Count of customers churn")
plt.show()
```

```python
plt.figure(figsize=(4,4))
gb=df.groupby('Churn').agg({'Churn':'count'})

plt.pie(gb['Churn'],labels=gb.index,autopct="%1.2f%%")
plt.title("Percentage of churned customers",fontsize=10)
plt.show()
gb
```

#from the given pie chart we can conculde that 26.54% of our customer have churned out not let's explore the reason behind it.


```python
plt.figure(figsize=(4,4))
sns.countplot(x="gender",data=df,hue="Churn")
plt.title("Churn by Gender")
plt.show()
```

```python
plt.figure(figsize=(4,4))
ax=sns.countplot(x="SeniorCitizen",data=df)
ax.bar_label(ax.containers[0])
plt.title("Count of Customer by SeniorCitizen")
plt.show()
```

```python

# Step 1: Create a crosstab of counts
ct = pd.crosstab(df['SeniorCitizen'], df['Churn'])

# Step 2: Convert counts to percentages of the total
ct_percent = ct.div(ct.sum(axis=1), axis=0) * 100

# Step 3: Plot stacked bar chart with orange and blue
ax = ct_percent.plot(
    kind='bar',
    stacked=True,
    figsize=(4,4),
       # orange for first category, blue for second
)

plt.title("Churn by SeniorCitizen (% of total)")
plt.ylabel("Percentage")
plt.xlabel("SeniorCitizen")

# Step 4: Add percentage labels
for p in ax.patches:
    width, height = p.get_width(), p.get_height()
    x, y = p.get_xy()
    if height > 0:  # avoid labeling empty bars
        ax.text(x + width/2, y + height/2, f"{height:.1f}%", 
                ha='center', va='center', color="black")

# Step 5: Adjust legend
plt.legend(title="Churn", labels=ct.columns, loc="best")

# Step 6: Add percentage scale on y-axis
plt.ylim(0, 100)                       # limit from 0 to 100
plt.yticks(range(0, 101, 10))          # ticks every 10%

plt.show()
```

```python
cols = [
    'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
    'TechSupport', 'StreamingTV', 'StreamingMovies'
]

fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.flatten()

for i, col in enumerate(cols):
    sns.countplot(x=col, data=df, hue="Churn", ax=axes[i])
    axes[i].set_title(f"Countplot of {col}")
    axes[i].set_ylabel("Count")
    axes[i].set_xlabel(col)

plt.tight_layout()
plt.show()
```

#This dataset captures customer subscriptions across phone, internet, and streaming services, along with add‑ons like security, backup, device protection, and tech support. It compares these service features against churn status, showing how usage patterns influence customer retention. Senior citizen and multiple line usage provide demographic context. Overall, it offers a clear view of service adoption and its relationship to churn behavior.
